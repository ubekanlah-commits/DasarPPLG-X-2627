```python
#@title 🛒 Aplikasi Kasir Kantin RPL
nama_pembeli = "Budi" #@param {type:"string"}
menu_makanan = "Nasi Goreng Spesial" #@param ["Nasi Goreng Spesial", "Mie Ayam Bakso", "Soto Ayam"]
harga_satuan = 15000 #@param {type:"integer"}
jumlah_porsi = 1 #@param {type:"slider", min:1, max:10, step:1}
uang_dibayar = 50000 #@param {type:"integer"}

# LENGKAPI RUMUS DI BAWAH INI:
total_belanja = harga_satuan * jumlah_porsi
uang_kembalian = uang_dibayar - total_belanja

# Menampilkan Struk Belanja
print("======================================")
print("         STRUK KANTIN SMK RPL         ")
print("======================================")
print(f"Nama Pembeli   : {nama_pembeli}")
print(f"Menu Pesanan   : {menu_makanan}")
print(f"Harga Satuan   : Rp{harga_satuan}")
print(f"Jumlah Porsi   : {jumlah_porsi} porsi")
print("--------------------------------------")
print(f"TOTAL BELANJA  : Rp{total_belanja}")
print(f"UANG TUNAI     : Rp{uang_dibayar}")
print(f"UANG KEMBALIAN : Rp{uang_kembalian}")
print("======================================")
```
